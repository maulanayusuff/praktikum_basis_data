```python
def coin_change_greedy(amount, coins): #fungsi hitung kembalian koin dengan metode greedy
    coins.sort(reverse=True) #mengurutkan pecahan koin dari yang paling besar
    result= [] #wadah kosong list untuk menyimpan koin yang terpilih
    for coin in coins: #looping memeriksa setiap pecahan koin yang tersedia
        while amount >= coin: #looping selama sisa uang masih cukup dengan koin saat ini
            amount -= coin #kurangi sisa uang dengan nilai koin yang diambil
            result.append(coin) #masukkan koin yang berhasil diambil ke dalam list hasil
    return result #kembalikan hasil daftar koin penukaran yang optimal

amount= 57 #menentukan jumlah uang nominal yang ingin ditukarkan
coins= [25,10,5,1] #daftar pilihan pecahan koin yang tersedia
change= coin_change_greedy(amount, coins) #menjalankan fungsi greedy untuk mencari koin kembalian
print("koin yang digunakan:", change) #mencetak daftar koin yang digunakan ke layar
```

```python
def fractional_knapsack(items, capacity): #fungsi menghitung keuntungan maksimal kantong dengan sistem pecahan
  items= sorted(items, key=lambda x: x[1]/x[0], reverse=True) #urutkan barang berdasarkan nilai per bobot tertinggi
  total_value= 0.0 #set nilai awal total keuntungan barang jadi nol
  for value, weight in items: #looping memeriksa setiap barang (nilai, bobot) yang tersedia
    if capacity >= weight: #jika kapasitas kantong masih muat untuk menampung seluruh barang
      capacity -= weight #kurangi kapasitas kantong dengan bobot barang tersebut
      total_value += value #tambahkan seluruh nilai keuntungan barang ke total
    else: #jika kapasitas kantong sudah tidak muat menampung seluruh barang
      total_value += value * capacity / weight #ambil pecahan barang sesuai sisa kapasitas kantong
      break #hentikan perulangan karena kantong sudah penuh
  return total_value #kembalikan hasil total keuntungan maksimal yang didapat

items= [(60,10),(100,20),(120,30)] #daftar barang berbentuk (nilai_keuntungan, bobot_barang)
capacity= 50 #menentukan kapasitas berat maksimal yang bisa ditampung kantong
max_value= fractional_knapsack(items, capacity) #menjalankan fungsi untuk mencari keuntungan maksimal
print("nilai maksimum yang dapat dibawa:", max_value) #mencetak total nilai keuntungan maksimal ke layar
```

```python
import heapq #mengambil modul antrean prioritas otomatis
def prim_mst(graph, start): #fungsi mencari total bobot pohon merentang minimum (mst)
  visited= set () #wadah kosong berbentuk set untuk mencatat node yang sudah terhubung
  min_heap= [(0, start)] #antrean prioritas awal berisi pasangan (bobot, node_mulai)
  total_weight= 0 #set awal total bobot mst menjadi nol

  while min_heap: #looping selama antrean prioritas masih ada isinya
    weight, node = heapq.heappop (min_heap) #ambil garis dengan bobot paling kecil dari antrean
    if node in visited: #jika node tersebut ternyata sudah pernah dikunjungi
      continue #abaikan dan langsung lanjut ke antrean berikutnya
    visited.add(node) #masukkan node baru ini ke dalam daftar visited
    total_weight += weight #tambahkan bobot garis terpilih ke total bobot mst
    for neighbor, edge_weight in graph[node]: #periksa semua tetangga beserta bobot garis dari node sekarang
      if neighbor not in visited: #jika node tetangga belum terhubung ke dalam mst
        heapq.heappush(min_heap, (edge_weight, neighbor)) #masukkan jalur tetangga baru ke antrean prioritas
  return total_weight #kembalikan hasil akhir berupa total seluruh bobot mst

graph= { #membuat struktur data graf tidak berarah beserta bobotnya
    'A': [('B', 2), ('C', 3)], #node A terhubung ke B (bobot 2) dan C (bobot 3)
    'B': [('A', 2), ('C', 1), ('D', 1)], #node B terhubung ke A (bobot 2), C (bobot 1), dan D (bobot 1)
    'C': [('A', 3), ('B', 1), ('D', 4)], #node C terhubung ke A (bobot 3), B (bobot 1), dan D (bobot 4)
    'D': [('B', 1), ('C', 4)] #node D terhubung ke B (bobot 1) dan C (bobot 4)
}

print("total bobot mst (prim):", prim_mst(graph, 'A')) #menjalankan fungsi prim dari node A dan mencetak hasilnya
```

```python
def kruskal_mst(edges, n_nodes): #fungsi mencari total bobot mst dengan algoritma kruskal
  parent= {i: i for i in range (n_nodes)} #setiap node awal diatur sebagai induk (parent) dirinya sendiri

  def find(x): #fungsi internal untuk mencari ketua kelompok (root) dari suatu node
    while parent [x] != x: #looping naik ke atas selama node bukan induk dari dirinya sendiri
      x= parent[x] #pindah ke node induk di atasnya
    return x #kembalikan node ketua kelompok yang ditemukan

  def union(x, y): #fungsi internal untuk menggabungkan dua kelompok node yang berbeda
    root_x= find(x) #cari ketua kelompok dari node x
    root_y= find(y) #cari ketua kelompok dari node y
    if root_x != root_y: #jika kedua node berasal dari kelompok yang berbeda
      parent[root_y]= root_x #gabungkan kelompok y ke dalam kelompok x
      return True #kembalikan true tanda penggabungan berhasil (tidak bikin siklus)
    return False #kembalikan false jika kelompoknya sama (bisa bikin siklus)

  edges.sort(key=lambda x: x) #urutkan semua garis berdasarkan bobotnya dari yang terkecil
  total_weight= 0 #set awal total bobot mst menjadi nol
  for u, v, weight in edges: #looping memeriksa setiap garis (node_asal, node_tujuan, bobot)
    if union(u, v): #jika kedua node sukses digabungkan tanpa membentuk siklus
      total_weight += weight #tambahkan bobot garis tersebut ke total bobot mst
  return total_weight #kembalikan hasil akhir berupa total seluruh bobot mst

edges = [(0, 1, 2), (0, 2, 3), (1, 2, 1), (1, 3, 1), (2, 3, 4)] #daftar garis graf dengan format (u, v, bobot)
print("total bobot mst (kruskal):", kruskal_mst (edges, 4)) #menjalankan fungsi kruskal untuk 4 node dan mencetak hasilnya
```
